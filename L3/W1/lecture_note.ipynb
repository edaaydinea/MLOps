# Hyperparameter tuning: searching for the best architecture

## Hyperparameter Tuning

#### 1. **Hyperparameter Tuning in ML Models**
   - **Model Parameters vs. Hyperparameters**:
     - *Model Parameters*: Learned from the training set (e.g., weights and biases).
     - *Hyperparameters*: Adjustable parameters set before training (e.g., learning rate, batch size). These affect the training process but are not learned from the data.
   - **Challenges**:
     - Hyperparameter tuning is crucial for optimal model performance.
     - The number of hyperparameters can be large, even in small models (e.g., DNNs).
     - Manual tuning is tedious and requires tracking experiments, making it inefficient.
   - **Impact**:
     - Proper tuning can significantly boost model performance.
   - **Tools for Hyperparameter Tuning**:
     - Open-source libraries automate hyperparameter tuning.
     - **Keras Tuner**: A popular tool offering methods like Random Search, Hyperband, and Bayesian Optimization.

#### 2. **Neural Architecture Search (NAS)**
   - **Overview**:
     - NAS automates the design of neural network architectures, finding optimal structures that can outperform manually designed models.
     - It covers a vast parameter space, making automation through tools like AutoML essential.
   - **Goal**:
     - The goal of NAS is to find the best possible architecture for a given problem, eliminating the need for manual architecture design.
   - **Relation to Hyperparameter Tuning**:
     - NAS can be seen as an extension of hyperparameter tuning but focuses on architecture design rather than just tuning predefined model settings.

#### 3. **AutoML**
   - **Purpose**:
     - AutoML automates the entire machine learning workflow, including data preprocessing, model selection, and hyperparameter tuning.
   - **Advantages**:
     - Makes machine learning accessible to non-experts by reducing manual intervention.
     - NAS and hyperparameter tuning are key components of AutoML frameworks.
   - **Cloud Integration**:
     - AutoML tools are often integrated with cloud platforms for scalable and efficient model training.

#### **Key Concepts to Remember**
   - **Hyperparameter Tuning**: A process of optimizing the hyperparameters before training to improve model performance.
   - **Keras Tuner**: A tool to simplify hyperparameter tuning, offering different search methods.
   - **NAS**: Automates the design of neural networks, crucial for finding optimal architectures in modern ML applications.
   - **AutoML**: Automates the end-to-end ML pipeline, including NAS and hyperparameter tuning, often integrated with cloud solutions.

This summary should serve as a foundational reference when you study hyperparameter tuning, NAS, and AutoML.


## Keras Autotuner Demo

```python
# Setting up libraries and dataset

import tensorflow as tf
from tensorflow import keras

mnist = tf.keras.datasets.mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0
```

```python
# Building the model
model = tf.keras.models.Sequential([
    # Flatten layer flattens the input, which is a 28x28 image, into a 1D array of 784 pixels
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    # Dense layer is a fully connected layer with 128 nodes
    tf.keras.layers.Dense(128, activation='relu'),
    # Dropout layer is used to prevent overfitting
    tf.keras.layers.Dropout(0.2),
    # Output layer is a fully connected layer with 10 nodes, each representing a digit from 0 to 9
    tf.keras.layers.Dense(10, activation='softmax')
])

# Compiling and training the model
model.compile(optimizer='adam', # Adam is a popular optimizer
              loss='sparse_categorical_crossentropy', # Sparse categorical crossentropy is used for classification problems
              metrics=['accuracy']) # Accuracy is used as the metric

# Training the model
model.fit(x_train, y_train, epochs=5)

# Evaluating the model
model.evaluate(x_test, y_test)
```

```python
# Parameters rational: if any
# Changing the number of nodes in the hidden layer to 512

model = tf.keras.models.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(512, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy', 
              metrics=['accuracy']) 

model.fit(x_train, y_train, epochs=5)
model.evaluate(x_test, y_test)
```

**Is this architecture optimal?**

- Do the model need more or less hidden units to perform well?
- How does model size affect the convergence speed?
- Is there any trade off between convergence speed, model size and accuracy?
- Search automation is the natural path to take
- Keras tuner built in search fuctionality.

```python
#Automated search with Keras tuner

#First, install the Keras tuner
# !pip install -q -U keras-tuner  # I installed it in my local environment via terminal

# Import Keras Tuner after it has been installed
import kerastuner as kt
```

```python
# Building model with iterative search

def model_builder(hp):
    # Building the model
    model = keras.Sequential()
    # Flatten layer flattens the input, which is a 28x28 image, into a 1D array of 784 pixels
    model.add(keras.layers.Flatten(input_shape=(28, 28)))
    
    # Tune the number of units in the first Dense layer
    hp_units = hp.Int('units', min_value=16, max_value=512, step=16)
    # Dense layer is a fully connected layer with 128 nodes
    model.add(keras.layers.Dense(units=hp_units, activation='relu'))
    # Dropout layer is used to prevent overfitting
    model.add(keras.layers.Dropout(0.2))
    # Output layer is a fully connected layer with 10 nodes, each representing a digit from 0 to 9
    model.add(keras.layers.Dense(10))
    
    
    # Tune the learning rate for the optimizer
    model.compile(optimizer='adam',
                  loss = "sparse_categorical_crossentropy",
                  metrics=['accuracy'])
    
    return model
```

```python
# Search strategy

# Instantiate the tuner to perform the hypertuning
tuner = kt.Hyperband(model_builder,
                     objective='val_accuracy',
                     max_epochs=10,
                     factor=3,
                     directory='my_dir',
                     project_name='intro_to_kt')
```

```python
# Callback configuration

stop_early = tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5)

# Perform the hypertuning
tuner.search(x_train, y_train, epochs=50, validation_split=0.2, callbacks=[stop_early])
```

## Understanding Search Spaces

1. **Definition of a Node**:
   - A node refers to a layer in a neural network, such as a convolution or pooling layer.
   - Nodes represent different types of layers, and connections between them define how the output of one layer is fed into another.

2. **Types of Search Spaces**:
   - **Macro Search Space**:
     - Involves searching for the best neural network architecture by exploring individual layers and their connections.
     - The NAS process builds the network layer by layer within this search space.
     - Architectures can be:
       - **Chain-Structured**: Simple, sequential stacking of layers.
       - **Complex**: Involves multiple branches and skip connections.
   - **Micro Search Space**:
     - Focuses on building the neural network using cells instead of individual layers.
     - **Cells**: Small sub-networks, which can be of different types, like:
       - **Normal Cell**: Maintains the same spatial dimensions.
       - **Reduction Cell**: Reduces the spatial dimensions (e.g., through pooling or striding).
     - Cells are stacked to form the final network architecture, which can be simple or involve multi-branch structures.
     - **Performance Advantage**: The micro search space approach has been shown to outperform the macro approach, likely due to its ability to explore and optimize at a higher abstraction level.

3. **Comparison**:
   - **Macro Search Space**:
     - Granular control over individual layers.
     - Builds networks layer by layer.
     - May lead to simpler or more complex structures.
   - **Micro Search Space**:
     - Operates at a higher level, using cells as building blocks.
     - Cells can be combined in various ways, providing flexibility and better performance.
     - Tends to offer a more efficient search and potentially better models.

### Key Takeaway
- **Macro Search Space** builds networks layer by layer, offering detailed control over architecture, but **Micro Search Space** leverages the power of predefined cells, often leading to better performance and more efficient searches in NAS.

## Search Strategies

### Neural Architecture Search (NAS) - Search Strategies

Neural Architecture Search (NAS) utilizes different strategies to explore the search space for the best model architecture. Here’s an overview of the most common strategies:

1. **Grid Search**:
   - **Description**: This is an exhaustive search method that explores every possible combination in the search space.
   - **Use Case**: Works well in small search spaces but becomes computationally expensive and inefficient in large search spaces.

2. **Random Search**:
   - **Description**: Randomly selects architectures from the search space. 
   - **Use Case**: More efficient than grid search, but can still be slow in larger search spaces. Random search can sometimes stumble upon good architectures, but it's not systematic.

3. **Bayesian Optimization**:
   - **Description**: A more sophisticated approach that assumes the performance of model architectures follows a probability distribution (commonly a Gaussian distribution). 
   - **How It Works**: Uses observations from tested architectures to update the probability distribution and guide the selection of the next architecture to try. It builds a model of the architecture’s performance and uses this model to make informed decisions about the next architectures to evaluate.
   - **Benefit**: Efficient in guiding the search towards better-performing architectures based on previous results.

4. **Evolutionary Algorithms**:
   - **Description**: Inspired by biological evolution, this approach generates an initial population of architectures, evaluates them, and selects the best performers (parents) to create a new generation through mutations and combinations (offspring).
   - **How It Works**: 
     - **Initial Population**: Randomly generated architectures.
     - **Selection**: Top-performing architectures are selected as parents.
     - **Mutation**: Random alterations (e.g., adding/removing layers, changing layer size) are introduced to create offspring.
     - **Replacement**: Poor-performing architectures are removed, and the process repeats.
   - **Benefit**: Allows exploration of diverse architectures and can gradually improve the population over time.

5. **Reinforcement Learning (RL)**:
   - **Description**: Uses an agent to take actions in an environment (the search space), with the goal of maximizing a reward. 
   - **How It Works**: 
     - The environment is the search space, and the reward function is the performance estimation strategy.
     - The agent (often modeled by a neural network) learns from feedback (rewards) to improve its choices.
     - Architectures are generated as variable-length strings (e.g., sequences of layers), and an RNN, known as the **controller**, can be used to generate these strings.
     - The child network, which is generated by the controller, is trained and evaluated. Its accuracy on the validation set is used as the reward for updating the controller.
     - Over time, the controller learns to favor architectures that yield better results, refining the search.
   - **Example**: This method has been successfully used to design networks for datasets like CIFAR-10 that rival the best human-designed architectures.

### Key Takeaways:
- **Grid and Random Search**: Simple and useful for smaller search spaces, but inefficient for larger ones.
- **Bayesian Optimization**: Builds a model of architecture performance to make informed decisions, making it more efficient for larger spaces.
- **Evolutionary Algorithms**: Leverage concepts of natural selection to gradually improve architectures, allowing for exploration and adaptation.
- **Reinforcement Learning**: Uses agents to learn and improve the search over time, showing success in complex tasks like image recognition.

Each search strategy has its strengths and weaknesses, and the choice depends on the complexity of the search space and the problem being solved.

## Measuring AutoML Efficacy

### Performance Estimation Strategies in Neural Architecture Search (NAS)

In NAS, accurate performance estimation of different architectures is crucial for effective search. However, evaluating each architecture fully can be computationally expensive. Here are some strategies to address this challenge:

1. **Validation Accuracy Measurement**:
   - **Description**: Directly measures the accuracy of each architecture on a validation set.
   - **Drawback**: Computationally intensive, particularly for large search spaces and complex networks, potentially taking several GPU days.

2. **Lower Fidelity Estimates**:
   - **Description**: Reduces the computational cost by simplifying the problem, such as training on a smaller subset of data, using lower resolution images, or reducing the number of filters and cells.
   - **Pros**: Speeds up the evaluation process.
   - **Cons**: Underestimates performance, and the relative ranking of architectures might not be accurate. Recent research suggests that lower fidelity estimates can misrepresent the actual performance.

3. **Learning Curve Extrapolation**:
   - **Description**: Assumes the ability to predict the learning curve for architectures based on partial training results.
   - **How It Works**: Uses a few iterations to build a learning curve for each architecture and extrapolates this to predict performance, terminating poorly performing architectures early.
   - **Progressive Neural Architecture Search**: An approach that trains a surrogate model to predict performance based on architectural properties, which can efficiently guide the search process.

4. **Weight Inheritance or Network Morphism**:
   - **Description**: Initializes new architectures using the weights of previously trained models, similar to transfer learning.
   - **How It Works**: Modifies the architecture while preserving the underlying function, allowing the network to retain knowledge from the parent model.
   - **Pros**: Reduces training time significantly (often to just a few GPU days), maintains high performance by building on existing models, and allows for flexible search spaces without a predefined size limit.

### Summary of Performance Estimation Strategies:
- **Validation Accuracy Measurement**: Most accurate but computationally expensive.
- **Lower Fidelity Estimates**: Faster but less precise; may distort performance ranking.
- **Learning Curve Extrapolation**: Efficient if accurate extrapolation methods are available; involves predicting performance based on early training.
- **Weight Inheritance/Network Morphism**: Efficient and effective; leverages existing model knowledge to reduce training time and maintain performance.

Each strategy has trade-offs between accuracy and computational cost. The choice of strategy depends on the specific requirements of the NAS process and the resources available.
