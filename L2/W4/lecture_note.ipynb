## Semi-supervised Learning

#### **1. Importance of Advanced Labeling Techniques**
- **Machine Learning Dependency on Labeled Data**: Supervised learning requires labeled training sets, but manual labeling is expensive and difficult.
- **Advanced Labeling Benefits**: Reduces labeling costs and leverages large amounts of unlabeled data.
- **Common Scenario**: Labeled data is expensive; unlabeled data is cheap and abundant.

#### **2. Semi-Supervised Labeling**
- **Overview**: Combines a small, labeled dataset with a large, unlabeled dataset. Labels for the unlabeled data are inferred based on the structure of the labeled data in the feature space.
- **Key Assumption**: Labeled data will cluster together or have recognizable structure in the feature space.
- **Advantages**:
  - Improves model accuracy by utilizing both labeled and unlabeled data.
  - Unlabeled data is inexpensive and easily accessible.
- **Label Propagation Algorithm**:
  - **Semi-Supervised**: Assigns labels to unlabeled data points based on the structure of the labeled data.
  - **How It Works**: Based on the similarity or community structure of labeled and unlabeled data points.
  - **Example**: Graph-based label propagation—labels are assigned to unlabeled examples based on neighbors and then propagated through the clusters.
  - **Transductive Learning**: Mapping is done directly from examples, without learning a general function for mapping.

#### **3. Active Learning**
- **Concept**: Uses intelligent sampling to assign labels to unlabeled data based on existing labeled data.
- **Goal**: Select the most informative examples to label, improving the model with fewer labeled instances.

#### **4. Weak Supervision**
- **Definition**: Programmatically assigning labels using heuristics designed by subject matter experts. It allows for the use of weaker, noisier labels at a larger scale.
- **Example Framework**: **Snorkel** - A friendly framework for applying weak supervision through labeling functions and aggregating noisy labels.
  
#### **5. Data Augmentation**
- **Purpose**: Increase the diversity of training data without manually collecting new data.
- **Common Techniques**: 
  - For images: rotation, flipping, cropping.
  - For text: synonym replacement, back-translation.
  - For audio: noise addition, time-stretching.

#### **6. Data Pre-processing**
- **Importance**: Clean, prepare, and structure data to improve model performance and training efficiency.
- **Common Steps**:
  - **Handling Missing Data**: Imputation, removal.
  - **Normalization/Standardization**: Scaling features to a consistent range or distribution.
  - **Categorical Encoding**: Converting categorical data into numerical forms (e.g., One-hot encoding, Label encoding).

### **Key Terms to Remember**
- **Semi-Supervised Learning**: Combines labeled and unlabeled data for training.
- **Label Propagation**: A semi-supervised algorithm that spreads labels through the feature space based on similarities.
- **Active Learning**: Prioritizing the labeling of the most informative data points.
- **Weak Supervision**: Assigning programmatic labels using heuristics instead of manually labeling everything.
- **Snorkel**: A framework for weak supervision and programmatically labeling data.
- **Data Augmentation**: Techniques to create more diverse training data without manual collection.
- **Data Pre-processing**: Techniques to clean and prepare data for model training, such as handling missing values, normalization, and encoding.


## Active Learning

#### **1. What is Active Learning?**
- **Definition**: Active learning is an intelligent sampling technique that selects the most informative unlabeled data points to label, optimizing the learning process with a limited labeling budget.
- **Key Goal**: Improve model performance by selecting the data points that bring the most predictive value, reducing the need for labeling large datasets.
  
#### **2. Importance of Active Learning**
- **Limited Data Budget**: Labeling data can be expensive, especially in fields like healthcare, where expert annotation is required. Active learning minimizes costs by focusing on the most valuable data points.
- **Imbalanced Datasets**: Active learning efficiently identifies rare classes, which helps balance the dataset during training.
- **Improving Accuracy**: In cases where standard sampling techniques fail to improve model accuracy, active learning provides a method to achieve better performance with fewer labeled examples.

#### **3. Active Learning Lifecycle**
1. **Start with a Pool of Unlabeled Data**: Begin with a large pool of unlabeled data.
2. **Intelligent Sampling**: Select the most informative examples from the pool using intelligent sampling techniques.
3. **Label the Data**: Human annotators or automated techniques label the selected data.
4. **Train the Model**: Use the newly labeled data to train a model and make predictions.
5. **Repeat**: Continue the cycle by selecting and labeling new data points until the model performance converges.

#### **4. Intelligent Sampling Techniques**
- **Margin Sampling**:
  - **Concept**: Focuses on labeling the most uncertain data points, typically those closest to the decision boundary in a classification problem.
  - **Process**: Label the most uncertain data point, retrain the model, and repeat until the model stops improving.
  - **Benefit**: Achieves higher accuracy with fewer labeled examples compared to random sampling.
  
- **Cluster-Based Sampling**:
  - **Concept**: Selects a diverse set of data points by clustering the feature space, ensuring the model learns from a broad representation of the data.
  
- **Query-By-Committee**:
  - **Concept**: Trains multiple models (the "committee") and selects data points with the highest disagreement among them for labeling.
  
- **Region-Based Sampling**:
  - **Concept**: Divides the input space into different regions and applies active learning algorithms within each region. This method is relatively new and focuses on improving model performance by regionally optimizing the labeled data.

#### **5. Comparison of Sampling Techniques**
- **Random Sampling**: Labels data points randomly. It eventually reaches similar performance as active learning methods but requires much more labeled data.
- **Margin Sampling**: Achieves higher accuracy faster than random sampling by focusing on uncertain points near the decision boundary.
  
#### **Key Terms to Remember**
- **Active Learning**: A technique to intelligently sample the most informative data points for labeling.
- **Intelligent Sampling**: Selection of data points that provide the most predictive value, improving model performance efficiently.
- **Margin Sampling**: Labeling data points near the decision boundary to reduce uncertainty in the model.
- **Cluster-Based Sampling**: Selecting diverse data points by clustering the feature space.
- **Query-By-Committee**: Selecting data points with high disagreement among multiple models.
- **Region-Based Sampling**: Running active learning within specific regions of the input space.


## Weak Supervision

#### **1. What is Weak Supervision?**
- **Definition**: Weak supervision is an approach to generating labels using noisy and imperfect information from one or more sources, often heuristics created by subject matter experts.
- **Goal**: To label data programmatically when manual labeling is costly or impractical. The key challenge is managing noisy labels with probabilistic correctness rather than deterministic certainty.

#### **2. Components of Weak Supervision**
- **Noisy Sources**: Conditional distributions over unlabeled data that provide probabilistic labels rather than deterministic ones.
- **Heuristics**: Rule-based labeling procedures created by subject matter experts. These heuristics provide noisy labels that cover a portion of the data.
- **Generative Model**: A model that learns to assess the trustworthiness of each weak supervision source, effectively determining the relevance of the noisy labels provided by different sources.
  
#### **3. Process of Weak Supervision**
1. **Start with Unlabeled Data**: Begin with a pool of data without known true labels.
2. **Apply Labeling Functions**: Use heuristics (labeling functions) created by subject matter experts to generate noisy labels for the data.
3. **Use a Generative Model**: Train a generative model to de-noise the noisy labels by estimating the reliability of each labeling function.
4. **Train a Discriminative Model**: Finally, train a machine learning model using the de-noised labels produced by the generative model.

#### **4. Example: Snorkel Framework**
- **Overview**: Snorkel, developed at Stanford in 2016, is a leading framework for implementing weak supervision. It programmatically builds and manages training datasets without requiring manual labeling.
- **Capabilities**:
  - **Labeling Functions**: Snorkel allows the creation of labeling functions (heuristics) that automatically label data. These functions might label data based on simple rules (e.g., marking messages as spam if they contain certain keywords).
  - **Generative Model**: Snorkel uses a generative model to aggregate and de-noise the labels generated by multiple labeling functions.
  - **Discriminative Model**: After de-noising the labels, Snorkel trains a final model using these improved labels.
  - **Data Augmentation and Slicing**: In addition to weak supervision, Snorkel offers tools for augmenting data and slicing datasets for better model performance.

#### **5. Example Labeling Functions in Snorkel**
- **Function 1**: Labels a message as spam if it contains the word "my." Otherwise, it returns "abstained" (indicating that it doesn't have enough information to label the data).
- **Function 2**: Labels a message as spam if it is longer than five words. 
- **Goal**: Use multiple labeling functions to improve the likelihood of correctly labeling the data.

#### **6. Key Points Recap**
- **Supervised Learning**: Relies on labeled data, but manual labeling can be expensive and time-consuming.
- **Semi-Supervised Learning**: Combines small amounts of labeled data with large amounts of unlabeled data to improve model accuracy.
- **Active Learning**: Uses intelligent sampling to select the most informative examples for labeling, improving accuracy while minimizing labeling costs.
- **Weak Supervision**: Uses noisy, limited, or inaccurate label sources to label data programmatically in a supervised learning environment.
- **Snorkel Framework**: A widely used system that facilitates weak supervision by managing labeling functions, de-noising labels, and training models.

### **Key Terms to Remember**
- **Weak Supervision**: Labeling data using noisy, probabilistic sources rather than deterministic labels.
- **Heuristics**: Rule-based labeling functions created by experts to generate noisy labels.
- **Generative Model**: A model that learns to evaluate the reliability of labeling functions and de-noises the labels.
- **Snorkel**: A framework for weak supervision that programmatically builds labeled datasets, de-noises labels, and supports data augmentation and slicing.


# Data Augmentation

#### **1. What is Data Augmentation?**
- **Definition**: Data augmentation involves expanding a dataset by creating slightly modified versions of existing data or generating new synthetic data. The goal is to improve model performance by increasing the diversity and coverage of the feature space.

#### **2. Why Use Data Augmentation?**
- **Enhances Model Performance**: By adding more valid examples, data augmentation improves feature space coverage, helping the model generalize better.
- **Reduces Overfitting**: Augmenting data makes it less likely that the model will overfit to specific patterns or rotations in the training data.
- **Cost-effective**: It provides an alternative to acquiring new labeled data, which can be expensive or time-consuming.

#### **3. Methods of Data Augmentation**
- **Simple Transformations**:
  - **Flips and Rotations**: Basic image manipulations like flipping, rotating, or padding/cropping images.
  - **Example**: Using the CIFAR-10 dataset, images can be augmented by padding, cropping, flipping, and rotating. TensorFlow functions like `tf.image.resize`, `tf.image.random_crop`, and `tf.image.random_flip_left_right` are used to perform these operations.
- **Advanced Techniques**:
  - **Unsupervised Data Augmentation (UDA)**: A method of augmenting data in a semi-supervised setting, where labeled and unlabeled data are used together.
  - **Policy-based Augmentation**:
    - **AutoAugment**: Automatically learns policies for augmenting data in ways that enhance model performance.
    - **Example**: For text data, a policy-based method could generate valid variants of a movie review, creating multiple new examples from one.

#### **4. Practical Example: CIFAR-10 Dataset**
- **Dataset Overview**: CIFAR-10 consists of 60,000 32x32 color images divided into 10 classes, with 6,000 images per class. It's widely used for machine learning research in computer vision.
- **Augmentation Example**:
  - **Padding and Cropping**: Padding an image and then cropping it back to the original size (32x32) adds variety to the dataset without changing the core content.
  - **Flipping**: Randomly flipping images left or right introduces more variations, preventing the model from overfitting to specific orientations.

#### **5. Benefits of Data Augmentation**
- **Increases Dataset Size and Diversity**: Augmenting the dataset by generating new examples increases both the quantity and variability of data, which leads to better model performance.
- **Improves Generalization**: By introducing slight variations, the model learns to generalize better to unseen data.
  
#### **6. Key Points Recap**
- **Data Augmentation**: A method to create additional labeled data by modifying existing data, thereby increasing feature space coverage and reducing overfitting.
- **Simple Techniques**: Include transformations like flips, rotations, and padding, often used for image data.
- **Advanced Techniques**: Include policy-based methods like AutoAugment, and unsupervised data augmentation approaches, which can be applied across various data types, including text.
- **Practical Example**: The CIFAR-10 dataset showcases basic image augmentation techniques such as padding, cropping, and flipping, which help increase dataset diversity.

### **Key Terms to Remember**
- **Data Augmentation**: Expanding datasets by generating new labeled examples from existing ones.
- **Feature Space Coverage**: Refers to how well the training data represents all possible scenarios that the model might encounter.
- **Overfitting**: A situation where a model performs well on training data but poorly on unseen data due to excessive focus on specific patterns in the training data.
- **AutoAugment**: A policy-based data augmentation method that automates the selection of optimal augmentation strategies.

# Preprocessing Different Data Types

## Time Series

#### **1. Types of Data**
- **Diverse Data Types**: Depending on your problem, you might work with various data types like images, videos, text, audio, or time series data. Each requires different preprocessing and modeling techniques.
- **Focus on Time Series Data**: Among these, time series data is often the least familiar to developers. This summary will focus on key aspects of time series data, particularly using sensor data for predictive modeling.

#### **2. What is Time Series Data?**
- **Definition**: Time series data is a sequence of data points recorded over time, typically at regular intervals. This type of data is used to make predictions about future values based on past measurements.
- **Example**: In weather prediction, historical data such as temperature, pressure, and humidity are used to forecast future conditions.

#### **3. Time Series Forecasting**
- **Goal**: The main objective of time series forecasting is to predict future values (e.g., temperature) based on past data.
- **Challenges**: Forecasting the future is inherently difficult due to uncertainties, while predicting the past is straightforward because it has already happened.

#### **4. Example: Weather Data**
- **Dataset**: The dataset from the Max Planck Institute for Biogeochemistry contains 14 features, including temperature, pressure, and humidity, recorded every 10 minutes since 2003.
- **Target Variable**: The goal is to predict the temperature, using other features as input.

#### **5. Data Characteristics**
- **Periodicity and Seasonality**: The data exhibits clear periodic patterns, likely due to seasonal changes. However, "seasonality" in time series data can refer to any kind of periodic pattern, not just annual seasons.
  
#### **6. Windowing Strategies**
- **What is Windowing?**: In time series forecasting, windowing involves creating time-based windows to capture dependencies in the data. The window size refers to the number of past observations used to predict future values.
- **Sliding Window Example**: Suppose you want to predict the temperature one hour into the future, given six hours of historical data. You could use a sliding window with a history size of six hours and a prediction window of one hour, resulting in a total window size of seven.
- **Avoiding Time Travel**: Ensure that the model does not include future data in the training set, which could introduce "time travel" and lead to misleading results.

#### **7. Sampling Strategies**
- **Observation Frequency**: In the weather dataset, observations are recorded every 10 minutes, leading to 144 observations per day.
- **Sampling and Aggregation**: Instead of using every 10-minute observation, you might sample less frequently, such as once per hour. Another option is to aggregate data, for example by taking the median of observations within each hour, reducing the data size and improving computational efficiency.
  
#### **8. Practical Application: Preprocessing and Windowing with TFX**
- **Preprocessing**: Use TensorFlow Extended (TFX) to preprocess the weather dataset, transforming it into time sequences suitable for training models like Long Short-Term Memory (LSTM) networks.
- **Windowing with TensorFlow**: The `TF.data.Dataset.window` function in TensorFlow can be used to group data into windows for model training.
- **TFRecord Format**: Save the transformed data in the TFRecord format, which is optimized for TensorFlow training pipelines.

#### **9. Key Points Recap**
- **Time Series Forecasting**: Predicts future values based on historical data.
- **Windowing**: A strategy to capture temporal dependencies by grouping past data points.
- **Avoiding Time Travel**: Ensure no future data is used in the training process.
- **Sampling and Aggregation**: Reduce data dimensionality by sampling or aggregating observations, making the dataset more manageable.
- **Tools**: Use TensorFlow and TFX to preprocess time series data, create windows, and prepare it for training in deep learning models.

### **Key Terms to Remember**
- **Time Series Data**: Data points collected over time at regular intervals.
- **Windowing**: A method to structure time series data into fixed intervals for model training.
- **Seasonality**: Periodic patterns in time series data, which may or may not correspond to the actual seasons of the year.
- **Sampling**: Selecting fewer data points (e.g., one observation per hour instead of every 10 minutes) to reduce dataset size.
- **TFX**: TensorFlow Extended, a platform for deploying production-ready machine learning pipelines.


## Sensors and Signals

#### **1. Defining Sensor-Generated Data**
- **Sensors and Signals**: A sensor, like an accelerometer found in smartphones, generates a sequence of data known as signals. These signals are time-indexed, making them a type of time series data, which is analyzed using time series techniques.
  
#### **2. Example Problem: Human Activity Recognition (HAR)**
- **Objective**: HAR aims to infer human activities (e.g., walking, sitting, running) from the patterns in accelerometer data.
  
#### **3. Challenges in Human Activity Recognition**
- **Segmentation**: Properly segmenting sensor data is crucial for accurately recognizing activities. This segmentation process divides the raw time-dependent acceleration data into smaller, more manageable segments. 
  - **Importance of Segmentation**: Similar to the windowing strategy used in time series analysis (like weather data), segmentation enables effective feature extraction, classification, and validation.
  - **Segment Length**: The length of these segments typically ranges from 1 to 10 seconds, depending on the context of the application and the sensor's sampling rate.

#### **4. Data Transformations for Modeling**
- **Transformations**: Once the data is segmented, it is transformed to improve model performance and accuracy. Several techniques can be used for this:
  - **Spectrogram**: A spectrogram is a representation of a signal's frequency content over time, which is useful for capturing the intensity of different data points in inertial data.
  - **Other Transformations**:
    - **Normalization and Encoding**: Standardizing data for consistency.
    - **Multichannel Processing**: Handling multiple data channels (e.g., X, Y, Z axes of accelerometer data).
    - **Fourier Transform**: Converts the time-domain signal into its frequency-domain representation.
