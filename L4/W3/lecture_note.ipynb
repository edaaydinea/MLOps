# ML Experiments Management and Workflow Automation

## Experiment Tracking

#### **1. Importance of Experiment Tracking**
- **Experimentation:** Machine learning (ML) is an experimental science, requiring systematic tracking of experiments to ensure reproducibility and meaningful progress.
- **Debugging:** ML debugging focuses on model convergence/generalization issues rather than traditional software bugs. Tracking experiments helps identify problems.
- **Efficiency:** Experiments, especially with large models/datasets and expensive accelerators, are time-consuming and costly. Effective tracking optimizes the use of resources.

#### **2. Core Elements of Experiment Tracking**
- **Reproducibility:** Track all factors necessary to reproduce results (e.g., code, hyperparameters, environment, metrics). Even small changes (e.g., learning rate adjustments) can significantly impact outcomes.
- **Comparison:** Systematic tracking allows meaningful comparison across experiments, guiding next steps.
- **Collaboration:** Shared tools and consistent tracking across teams facilitate better communication and results sharing.

#### **3. Experiment Tracking Tools**
- **Basic Tracking:** Free-form notes may work initially but become unwieldy as experiments grow.
- **Notebooks:** Widely used for ML experimentation but not directly suited for production due to magics, visualization code, etc.
  - **Tools for Notebook Management:**
    - **nbconvert:** Extract Python code from Jupyter notebooks.
    - **nbdime:** Diff and merge Jupyter notebooks.
    - **jupytext:** Convert/sync notebooks with Python files.
    - **neptune notebooks:** Version, diff, and share notebooks.
  
#### **4. Transition from Notebooks to Production**
- **Modular Code:** As experiments scale, shift from monolithic code to modular code for reusability, robustness, and maintainability.
- **Directory Hierarchy:** Useful for simple projects; evolves into code repositories and versioning (commits) for larger, distributed workflows.
- **Shared Monorepos:** In team settings, separate experiment tracking to avoid conflicts with shared code.

#### **5. Hyperparameter and Configuration Tracking**
- **Configuration Files:** Simple and robust for tracking hyperparameters. Versioned alongside code.
- **Command-Line Parameters:** Require code to save runtime parameters and associate them with experiments for easier analysis and visualization.
- **Hybrid Approach:** Use configuration files while saving parameters in a data store for the best of both worlds.

#### **6. Example of Saving Runtime Parameters**
- **Neptune AI API:** Illustrates how to save runtime parameter values using code, ensuring they are accessible for further analysis.

This summary provides key points for studying experiment tracking, an essential component of ML workflows in production.

## Tools for Experiment Tracking

#### **1. Importance of Data Versioning**
- **Why Version Data:** Just as you version code, models, and hyperparameters, data must also be versioned to ensure experiments are reproducible. Changes in feature vectors (adding, deleting, or modifying features) need to be tracked.
- **Reproducibility:** Versioning data ensures you can reproduce and compare results accurately, which is essential for experimentation and debugging in ML.

#### **2. Tools for Data Versioning**
- **Neptune:** Offers data versioning, experiment tracking, and a model registry.
- **Pachyderm:** Provides continuous data updates in the master branch while allowing experimentation with specific data commits in separate branches.
- **Delta Lake:** Runs on top of existing data lakes, offering data versioning, rollbacks, and full historical audit trails.
- **Git LFS:** Extends Git for handling large files (e.g., audio, videos, datasets) by replacing them with text pointers inside Git.
- **DoIt:** A SQL database that can be forked, cloned, branched, merged, pushed, and pulled, similar to a Git repository.
- **LakeFS:** An open-source platform that provides Git-like branching and committing models for data at petabyte scale.
- **DVC (Data Version Control):** An open-source version control system for ML projects that runs on top of Git.
- **ML-Metadata (MLMD):** A library for recording and retrieving metadata related to ML workflows, integrated with TFX but usable independently.

#### **3. Experiment Tracking in ML**
- **Tracking Experiments:** Experiment tracking enables developers to meaningfully compare results across different experiments and avoid repeating mistakes. Tag experiments with consistent labels and add notes to maintain organization.
- **Tools for Tracking:**
  - **TensorBoard:** A powerful tool for visualizing training metrics, such as confusion matrices, in real-time. It provides insights into model performance throughout training.
  - **Cloud-Based Tools:** Vertex TensorBoard and other cloud-based tools provide managed services, security, privacy, and persistent shareable links to dashboards for collaboration.
  - **Neptune AI:** Offers experiment management and sharing capabilities, allowing teams to track progress, review results, and inspire new ideas.

#### **4. Collaborative Experiment Tracking**
- **Organizing Results:** From the start, organize experimental results to facilitate revisiting and sharing them with your team. Tag experiments and add notes to avoid redundant work and make collaboration easier.
- **Managed Services:** Cloud-based services offer advantages like compliance and easy sharing of dashboards with persistent links, which enhances team collaboration on larger projects.

#### **5. Iterative Experimentation and Creative Iterations**
- **Baseline Approach:** Establish a baseline with achievable goals aligned with business objectives. Evaluate the baseline to gain insights and decide whether to pursue more complex modeling.
- **Balancing Metrics:** While ML tooling often focuses on modeling metrics, it’s crucial to also meet business goals for latency, cost, fairness, privacy, and regulatory compliance (e.g., GDPR).

This summary highlights key tools and strategies for data versioning and experiment tracking, emphasizing collaboration and efficiency in ML workflows.

## Introduction to MLOps

This section focuses on MLOps, the convergence of data science, machine learning, and software engineering practices to streamline the development, deployment, and management of machine learning models in production. Key takeaways include:

1. **Roles of Data Scientists and Software Engineers**:
   - Data scientists typically focus on model accuracy, feature engineering, and prototyping in notebooks.
   - Software engineers prioritize building scalable, maintainable, and secure products with a focus on performance, customer satisfaction, and stability.
   - Combining these roles is essential for production ML, where models need to deliver business value, considering cost, latency, fairness, and privacy.

2. **Challenges in ML Engineering**:
   - Unlike traditional software, ML models evolve with data changes. Ensuring reproducibility, tracking, and collaboration in ML projects is challenging due to the lack of established tools and processes in many teams.

3. **MLOps as a Discipline**:
   - MLOps aims to unify ML system development (Dev) and operation (Ops), applying DevOps principles to ML workflows.
   - Core MLOps practices include continuous integration (testing code, data schemas, and models) and continuous delivery (automatically deploying models and pipelines).
   - Continuous training is unique to ML systems, where models are retrained as new data arrives, ensuring they remain relevant and accurate.

4. **Phases of the ML Lifecycle**:
   - Data scientists and ML engineers typically begin by shaping data and developing models. The workflow progresses through continuous training, model deployment, and continuous monitoring.
   - MLOps ensures that the transition from development to production is smooth, with continuous updates to meet changing business needs and data conditions.

5. **Key MLOps Capabilities**:
   - **Model Versioning**: Creating and maintaining versions of models to address changing requirements and data.
   - **Monitoring**: Continuous tracking of model performance to detect decay or issues in real-time.
   - **Governance**: Ensuring compliance, access control, and auditing through detailed logs and metadata management.
   - **Security**: Protecting models from attacks, unauthorized access, and corruption by malicious data or denial of service.

6. **MLOps Platforms**:
   - These platforms provide tools for managing the ML lifecycle, including model discovery solutions that catalog and track metadata, architecture, and history, making models easily discoverable and shareable within teams.

This summary highlights how MLOps builds on software engineering and DevOps principles to address the unique challenges of managing machine learning models in production environments.

# MLOps Methodology

## MLOps Level 0

MLOps maturity refers to the level of automation and sophistication in managing the end-to-end lifecycle of machine learning (ML) systems. The goal is to automate the processes of data preparation, model training, deployment, and monitoring to ensure that models continuously perform well in production environments. Here's a breakdown of MLOps maturity levels, starting with the basic level:

### **Level 0: Manual Processes**
- **Process Overview**: At this level, ML processes are manual and often driven by scripts or notebooks. Data scientists manage the workflow by manually transitioning from one step to the next (e.g., data preparation, model training, validation).
- **Key Characteristics**:
  - **No Automation**: Every step, from data analysis to deployment, is manual.
  - **Experimental Code**: Data scientists work in notebooks, which can result in inconsistencies between development and production environments.
  - **Manual Handoffs**: Models are handed over to engineers manually for deployment, leading to possible training-serving skew.
  - **Low Frequency of Change**: Models are rarely retrained or updated, and new versions are deployed infrequently.
  - **No Continuous Integration/Continuous Deployment (CI/CD)**: Testing and deployment are not automated, and logs are minimal.
  - **Limited Monitoring**: There is no active monitoring for model decay or performance degradation in production.
  - **Challenges**: Without proper automation, models can break when deployed, leading to failures when the environment or data changes.

### **Key Focus at Level 0**:
- **Active Monitoring**: Introducing active performance monitoring is critical to detect performance issues and model decay.
- **Frequent Retraining**: Implementing mechanisms for continuous retraining with recent data can help adapt models to changing trends.
- **Manual Processes for Data and Labeling**: New data collection and labeling are still manual processes at this level.

As MLOps processes evolve, the goal is to increase automation, implement continuous integration, and ensure robust model monitoring and retraining mechanisms. This leads to higher levels of maturity, where ML systems are more resilient, scalable, and able to adapt to real-world changes with minimal manual intervention.

## MLOps Levels 1 & 2

**MLOps Level 1: Pipeline Automation**
- **Continuous Training**: Automating the training pipeline enables continuous delivery of models to the prediction service. New data is used to retrain models, triggered by specific events or schedules.
- **Pipeline Automation Characteristics**: Steps in experimentation are orchestrated, leading to faster iteration and easier transition to production. Components are reusable, containerized, and modularized for reproducibility across different environments (e.g., Dev, Test, Production).
- **Data and Model Validation**: Before model training, data validation is essential to ensure the input data is valid, preventing schema and value skews. After training, model validation compares the new model's performance with the current model to ensure improvements.
- **Feature Store**: A centralized repository for feature management, ensuring consistency between training and serving phases. It reduces redundant feature creation and prevents training-serving skew.
- **Metadata Store**: Stores execution information for data lineage, reproducibility, and error debugging. It helps resume pipeline execution from the last successful step in case of interruptions.

**MLOps Level 2: CI/CD for ML Pipelines**
- **Pipeline Architecture**: Focuses on rapid and reliable updates to production pipelines. This involves robust CI/CD to explore and implement new ideas efficiently.
- **Core Components**: Level 2 architecture includes source code control, build services, deployment services, model registry, feature store, metadata store, and pipeline orchestration.
- **Stages in ML CI/CD Lifecycle**:
  1. **Experimentation and Development**: Iteratively try new algorithms and modeling techniques.
  2. **CI/CD for Training Pipeline**: Build and test the source code for pipeline components, producing artifacts for deployment.
  3. **Model Training**: Execute the pipeline automatically in production, generating a trained model stored in a model registry.
  4. **Model Deployment**: Continuous delivery of trained models as a prediction service.
  5. **Monitoring**: Collect data on model performance, feeding back into the pipeline for retraining based on live data.

This level enables continuous integration of new features and ideas, improving the ML pipeline's robustness and flexibility in production environments.

## Developing Components for an Orchestrated Workflow

This lecture explains the process of developing training pipelines using TFX (TensorFlow Extended), emphasizing how to customize pipelines with custom components to meet specific needs. Here's a breakdown of the key points:

1. **Introduction to TFX**: 
   - TFX is an open-source framework that allows for the creation of ML pipelines.
   - TFX pipelines organize your ML workflow into components, where each component performs a specific task.
   - The framework supports different environments, including containerized environments like Kubernetes.
   - TFX standard components offer proven functionality, and custom components can be integrated to extend the pipeline.

2. **TFX Pipelines**:
   - A typical TFX pipeline includes both training and inference components.
   - Components are combined to build a pipeline that meets your specific workflow needs. You can mix standard and custom components to optimize the pipeline.

3. **Component Implementation**:
   - TFX components are implemented as a combination of three classes:
     - **Component Specification**: Defines the input-output contract for the component.
     - **Executor Class**: Implements the logic of the component.
     - **Component Class**: Packages the specification and executor into a component that can be used in the pipeline.

4. **Types of Custom Components**:
   - **Fully Custom Components**: These allow for deep customization by defining the component specification, executor, and component interface classes.
   - **Python Function-Based Custom Components**: These are the easiest to create and only require a Python function decorated with annotations.
   - **Container-Based Custom Components**: These allow you to integrate code written in any language, as long as it can be executed in a Docker container.

5. **Python Function-Based Components**:
   - This style simplifies component creation by letting you define a function with type hints and a decorator to specify inputs, outputs, and parameters.
   - It’s ideal for simple tasks like model validation.

6. **Container-Based Components**:
   - These are similar to creating Docker files.
   - You define a container image with your component’s dependencies and a command line as the entry point.
   - This approach is suitable for non-Python code or complex runtime environments.

7. **Fully Custom Components**:
   - This style is the most complex but offers the most flexibility. It allows you to define custom inputs, outputs, and parameters, and extend existing components by overriding the executor class.
   - It involves defining a component specification, an executor class (which performs the actual work), and a component class (to combine the two and integrate them into the pipeline).

8. **Integrating Custom Components into Pipelines**:
   - After creating a custom component, you integrate it into the pipeline by wiring it to upstream and downstream components.
   - The new component is then added to the pipeline's components list.

This comprehensive guide to TFX component development shows how to leverage both standard and custom components to build adaptable ML workflows.

# Model Management and Deployment Infrastructure

## Managing Model Versions

Version control is essential for managing machine learning models because, just like in software development, it supports collaboration, rollback, continuous integration, and more. Model generation is an iterative process, and keeping track of various models, data, configurations, and code is crucial for reproducing results, improving performance, and enabling team collaboration.

Here are some key points on model versioning:

### Model Versioning:
1. **Why It's Important**:
   - **Collaboration**: Allows multiple developers to stay in sync.
   - **Rollback**: Facilitates reverting to previous versions when issues arise.
   - **Continuous Integration**: Supports automation of testing and integration processes.

2. **Challenges**:
   - **Complexity**: Models involve multiple components like code, data, and configurations.
   - **Reproducibility**: Ensuring that previous versions of the model can be accurately reproduced.

3. **Versioning Approaches**:
   - Similar to software, you can use a combination of three numbers to denote versions: **major**, **minor**, and **pipeline** versions:
     - **Major Version**: Incremented for incompatible data changes (e.g., schema changes).
     - **Minor Version**: Incremented for enhancements in model output.
     - **Pipeline Version**: Incremented for changes in the training pipeline, even if the model itself remains unchanged.
   - Other approaches include arbitrary grouping, black-box functional models, and pipeline execution versioning.

### Model Lineage:
- **Model Lineage** tracks relationships among artifacts that contributed to the trained model, including code, data, and preprocessing operations. It is crucial for recreating different model versions and understanding the model's history.

### Model Registries:
- **Model Registry**: A central repository for storing trained models, which helps with:
  - **Model Discovery**: Facilitates finding and reusing models.
  - **Collaboration**: Enables sharing models across teams.
  - **Governance**: Enforces guidelines for model uploads.
  - **Streamlined Deployment**: Supports continuous evaluation, monitoring, and promotion of models.

- **Metadata in Model Registries**: Includes model versions, serialized artifacts, and annotations that promote discoverability and understanding. Some registries also link to other artifacts and metadata stores to support model lineage.

### Popular Model Registries:
Examples of model registries include:
- Azure ML Model Registry
- SAS Model Manager
- MLflow Model Registry
- Google AI Platform
- Algorithmia

These registries vary in their features but all contribute to managing models effectively in production environments.

## Continuous Delivery

In a mature MLOps process, where managing numerous models is essential, implementing a robust deployment process is crucial, especially when model predictions are part of a user-facing application. This is where **continuous delivery (CD)** comes into play, ensuring smooth transitions from development to production.

### Continuous Integration (CI) and Continuous Delivery (CD)

**Continuous Integration (CI)** is the process of automatically building, packaging, and testing code as soon as new code is committed to the repository. In machine learning, CI not only focuses on traditional software testing (e.g., unit tests for code), but also includes tests for data and models.

- **Unit Testing for Data**: Ensures that feature engineering is correct, checking that features are scaled, normalized, or one-hot encoded properly, and that data types, ranges, and non-null values are as expected.
- **Unit Testing for Models**: Ensures that model outputs have the correct shape, handle NaN values appropriately, and that performance metrics like accuracy or AUC meet specified baselines.

**Continuous Delivery (CD)** is the automation of deploying new code and models to production environments. It ensures that the new model is compatible with the target environment and performs as expected in production. The CD process typically includes checks for prediction service performance and the model's ability to handle live data.

### Key Aspects of CI/CD for Machine Learning:

1. **CI/CD Workflow**:
   - **CI** triggers when new code is committed, automating build, package, and test processes.
   - If all tests pass, the code and models are promoted to a **CD pipeline**, where they are deployed to production environments, such as an online serving environment or a batch prediction service.

2. **Model Performance Monitoring**:
   - During serving, performance metrics are continuously collected to ensure the model performs well on live data. Continuous evaluation and monitoring are essential for detecting drifts in model performance over time.

3. **Infrastructure Validation**:
   - Infrastructure validation serves as an early warning layer before pushing a model to production. It focuses on ensuring that the model can be successfully loaded and run by the model server in the production environment.
   - In a TFX pipeline, the `infraValidator` component launches a sandbox model server with the model to verify compatibility. If the model behaves as expected, it is "blessed" and ready for deployment.

4. **Mock Data and Edge Cases**:
   - Mock datasets should be designed to cover edge and corner cases in the model’s feature space. These datasets should be much smaller than actual data but representative of critical cases.
   - Good mocks and unit tests help ensure robust code coverage, and libraries are available to track and test that coverage.

5. **Data and Model Analysis**:
   - During experimentation, data and model analysis are often manual. However, in a production pipeline, these analyses are automated, which ensures that the model is continuously evaluated and refined based on new data.

### Benefits of CI/CD for ML:
- **Automation**: Streamlines the deployment of models and ensures that models are tested, validated, and monitored continuously.
- **Reliability**: Helps prevent errors during deployment, reducing the risk of deploying a model that might fail in production.
- **Scalability**: Enables the management of a large number of models and deployments in an organized and scalable way.
- **Governance**: Enforces approval guidelines, ensuring that models meet the necessary standards before being deployed.

By integrating continuous delivery into your MLOps pipeline, you ensure that model deployments are smooth, reliable, and scalable, with ongoing monitoring to maintain model performance.

## Progressive Delivery

**Progressive delivery** is an advanced deployment strategy that builds on continuous integration (CI) and continuous delivery (CD). Its primary goal is to deliver updates and new features gradually, mitigating risk and allowing for better assessment of changes before they are fully deployed. Progressive delivery techniques include **canary deployments**, **A/B testing**, **multi-armed bandits**, and **contextual bandits**. Let’s explore these methods and their applications in machine learning deployments.

### Key Concepts of Progressive Delivery

1. **Blue/Green Deployment**:
   - Two environments, blue (current live) and green (staging/testing), are used. After deploying to green and performing tests, traffic is gradually switched from blue to green. This method provides zero-downtime deployment, easy rollback, and allows for thorough testing before going live.

2. **Canary Deployment**:
   - Canary deployment is a more gradual rollout compared to blue/green. Instead of switching all traffic at once, a small percentage of traffic is routed to the new version (green), and performance is monitored. Based on the results, traffic can either be increased to the new version or rolled back. This minimizes exposure to potential issues, making it a safer approach.

3. **Live Experimentation**:
   - Progressive deployment is closely linked to **live experimentation**, where models are tested in a production-like environment with real user data. Unlike traditional offline validation (e.g., accuracy or AUC-ROC), live experimentation evaluates models based on their business impact (e.g., profit, user engagement).

4. **A/B Testing**:
   - **A/B testing** involves comparing two or more models by splitting users into groups and routing each group to a specific model. The goal is to measure the impact on key business metrics. For example, in a recommendation system, different models might optimize for click-through rate or profit. Each user's session is consistently served by the same model to ensure valid comparisons. The winning model is selected based on performance.

5. **Multi-Armed Bandits**:
   - The **multi-armed bandit** approach improves on A/B testing by dynamically adjusting the traffic routed to each model as performance data is gathered. Models that perform better receive more traffic, while underperforming models are phased out more quickly. This approach leverages **reinforcement learning** to balance exploration (trying different models) and exploitation (using the best-performing models).

6. **Contextual Bandits**:
   - **Contextual bandits** extend the multi-armed bandit approach by incorporating context or environment into the decision-making process. For instance, in a recommendation system, the model might consider the user's location, time of day, or other contextual factors when determining which model to serve. This helps the system adapt to different situations and further reduces regret (the difference between optimal and actual performance over time).

### Benefits of Progressive Delivery

- **Reduced Risk**: By rolling out updates gradually and using techniques like canary deployments, you can catch issues early before they affect all users.
- **Improved Feedback**: Progressive delivery allows for real-time feedback and monitoring, enabling you to validate the impact of changes on actual business metrics.
- **Faster Iterations**: With strategies like A/B testing and multi-armed bandits, you can quickly iterate and optimize models based on live data, leading to faster and more reliable deployments.

### Progressive Delivery in ML Pipelines

When integrating progressive delivery into machine learning pipelines, the following steps are typically involved:

1. **Develop and Train Models**: Multiple candidate models are trained and validated using offline metrics.
2. **Deploy to Staging/Shadow Environment**: Models are deployed to a non-production environment where they can be tested in isolation, e.g., using blue/green or canary deployment.
3. **Evaluate in Live Experiments**: Models are exposed to real user data through techniques like A/B testing or multi-armed bandits. Performance is monitored, and the best model is selected based on business outcomes.
4. **Full Rollout**: Once the best-performing model is identified, it is gradually rolled out to all users using canary deployment or similar techniques.

### Summary

Progressive delivery takes deployment strategies a step further by enabling safer and more gradual rollouts. It allows for better control, reduced risk, and more dynamic optimization of models in production. Techniques like blue/green deployments, canary deployments, A/B testing, and multi-armed bandits help to ensure that new models are thoroughly tested and optimized before fully replacing existing systems.
